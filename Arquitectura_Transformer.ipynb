## 2) Visualización de las Capas y Hidden States 🔍

````python
from transformers import BertTokenizer
import torch
````

Cargar tokenizer y modelo BERT en español

````python
tokenizer = BertTokenizer.from_pretrained("dccuchile/bert-base-spanish-wwm-cased")
model = BertModel.from_pretrained("dccuchile/bert-base-spanish-wwm-cased", output_hidden_states=True)
````

Texto de ejemplo

````python
text = "El producto de Samsung Galaxy S21 llegó el 12 de marzo y superó mis expectativas."
inputs = tokenizer(text, return_tensors="pt", truncation=True, max_length=128)
outputs = model(**inputs)
````

````python
print(outputs)
````

Extraer los hidden states


````python
hidden_states = outputs.hidden_states
print("Número de capas (incluyendo la capa de embedding):", len(hidden_states))
````

Por ejemplo, si BERT base tiene 12 capas, se imprimirá 13 (1 de embedding + 12 codificadores)

## Análisis Sección 1

1. Módulos principales de BertModel (según print(model)):
- embeddings (BertEmbeddings): word_embeddings (vocabulario de 31002 tokens), position_embeddings (512 posiciones) y token_type_embeddings (2 segmentos), todos de dimensión 768, seguidos de LayerNorm y dropout.
- encoder (BertEncoder): 12 capas BertLayer apiladas (0-11), cada una con atención multi-cabeza y red feed-forward.
- pooler (BertPooler): Linear(768→768) con activación Tanh, aplicada al token [CLS].

2. Subcapas dentro de una BertLayer:
- Mecanismo de atención: query, key y value (Linear 768→768) dentro de BertSelfAttention.
- Residual y normalización: BertSelfOutput (dense + dropout + LayerNorm, sumado a la entrada). La misma estructura se repite en BertOutput, después de la feed-forward (BertIntermediate).

3. Dimensión oculta: hidden_size (d_model) = 768, la de las capas lineales de query, key y value.

## Análisis Sección 2

1. input_ids: IDs del vocabulario WordPiece de los 22 tokens: [CLS] (ID 4) al inicio y [SEP] (ID 5) al final. Palabras como "Samsung", "Galaxy" y "S21" se dividen en subpalabras (Sam ##su ##ng, Gala ##xy, S ##21). token_type_ids: indica el segmento (0 = oración A, 1 = oración B); son todos 0 porque hay una sola oración. attention_mask: son todos 1 porque no hay padding; los 22 tokens son reales y deben ser atendidos.

2. len(hidden_states) = 13: la posición 0 es la salida de la capa de embeddings (palabra + posición + segmento, con LayerNorm), antes del encoder; las posiciones 1 a 12 son la salida de cada una de las 12 BertLayer.

3. hidden_states[0].shape = torch.Size([1, 22, 768]): 1 = batch_size (una oración), 22 = sequence_length (tokens con [CLS] y [SEP]), 768 = hidden_size (vector por token).

## Sección 3: Pseudocódigo de Self-Attention

Fórmula: Attention(Q,K,V) = softmax(QKᵀ / √dk) · V

```
Entrada: X [n, 768] (un vector por token)

1. Q = X·Wq ; K = X·Wk ; V = X·Wv   (proyecciones lineales)
2. S = Q·Kᵀ                          (similitud de cada token con todos)
3. S = S / √dk                       (escala; dk = 64 por cabeza)
4. A = softmax(S) por fila           (pesos que suman 1)
5. Z = A·V                           (suma ponderada de los valores)
6. Multi-cabeza: repetir en 12 cabezas, concatenar y proyectar con Wo
```

### Relación con BertSelfAttention

El diagrama y el pseudocódigo van al cierre porque resumen lo que hacen por dentro los módulos de la Sección 1. `BertSelfAttention` contiene los `Linear` query, key y value (pasos 1 a 5), y `BertSelfOutput` aplica la proyección final, el residual y LayerNorm. Los `hidden_states` de la Sección 2 son el resultado de repetir este cálculo capa por capa.
